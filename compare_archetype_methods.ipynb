# Comparación: BERTopic vs Claude para Descubrimiento de Arquetipos

Este notebook compara dos enfoques para descubrir arquetipos:

1. **BERTopic con Threads**: Clustering automático basado en similitud semántica
2. **Claude Sonnet 4.5**: Análisis de lenguaje natural con comprensión contextual

````python
import json
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np

print("✓ Librerías importadas")
````

## 1. Cargar Resultados de Ambos Métodos

````python
# Cargar resultados BERTopic
try:
    with open('discovered_archetypes_from_threads.json', 'r', encoding='utf-8') as f:
        bertopic_results = json.load(f)
    print('✓ Cargados resultados de BERTopic')
    print(f"  - Fecha: {bertopic_results['discovery_date'][:10]}")
    print(f"  - Threads analizados: {bertopic_results['total_threads_analyzed']}")
    print(f"  - Silhouette Score: {bertopic_results['silhouette_score']:.3f}")
    print(f"  - Arquetipos encontrados: {len(bertopic_results['archetypes'])}")
    bertopic_available = True
except FileNotFoundError:
    print('❌ No se encontró discovered_archetypes_from_threads.json')
    print('   Ejecuta: python discover_archetypes_with_threads.py')
    bertopic_available = False

print()

# Cargar resultados Claude
try:
    with open('claude_archetype_suggestions.json', 'r', encoding='utf-8') as f:
        claude_results = json.load(f)
    print('✓ Cargados resultados de Claude')
    print(f"  - Fecha: {claude_results['analysis_date'][:10]}")
    print(f"  - Muestra analizada: {claude_results['sample_size']} mensajes")
    print(f"  - Arquetipos sugeridos: {len(claude_results['suggested_archetypes'])}")
    claude_available = True
except FileNotFoundError:
    print('❌ No se encontró claude_archetype_suggestions.json')
    print('   Ejecuta: python claude_archetype_analysis.py')
    claude_available = False
````

## 2. Comparación de Alto Nivel

````python
if bertopic_available and claude_available:
    comparison_data = [
        ['Método', 'BERTopic', 'Claude Sonnet 4.5'],
        ['Arquetipos encontrados', len(bertopic_results['archetypes']), len(claude_results['suggested_archetypes'])],
        ['Basado en', 'Similitud semántica', 'Comprensión de contexto'],
        ['Datos analizados', f"{bertopic_results['total_threads_analyzed']} threads", f"{claude_results['sample_size']} mensajes muestra"],
        ['Costo', 'Gratis', '~$0.50 por análisis'],
        ['Velocidad', '~5-10 min', '~1-2 min'],
        ['Silhouette Score', f"{bertopic_results['silhouette_score']:.3f}", 'N/A (no usa clustering)'],
    ]

    df_comparison = pd.DataFrame(comparison_data[1:], columns=comparison_data[0])

    print('\n' + '='*80)
    print('📊 COMPARACIÓN DE MÉTODOS')
    print('='*80)
    print(df_comparison.to_string(index=False))
    print('='*80)
````

## 3. Arquetipos de BERTopic

````python
if bertopic_available:
    print('\n🤖 ARQUETIPOS DESCUBIERTOS POR BERTOPIC:\n')
    print('='*80)

    for i, arch in enumerate(bertopic_results['archetypes'], 1):
        print(f"\n{i}. {arch['name']}")
        print(f"   Threads: {arch['thread_count']}")
        print(f"   Keywords: {', '.join(arch['keywords'][:6])}")
        print(f"   {arch['description']}")
        print('-'*80)
````

## 4. Arquetipos de Claude

````python
if claude_available:
    print('\n🧠 ARQUETIPOS SUGERIDOS POR CLAUDE:\n')
    print('='*80)

    for i, arch in enumerate(claude_results['suggested_archetypes'], 1):
        print(f"\n{i}. {arch['nombre']} (~{arch.get('porcentaje_estimado', 'N/A')}%)")
        print(f"   Descripción: {arch['descripcion']}")
        print(f"   Keywords: {', '.join(arch['keywords'][:6])}")
        if 'ejemplos' in arch:
            print(f"   Ejemplos:")
            for ej in arch['ejemplos'][:2]:
                print(f"     - {ej[:100]}...")
        print('-'*80)

    print(f"\n💡 Observaciones de Claude:")
    print(claude_results['observations'])
````

## 5. Análisis de Solapamiento de Keywords

````python
if bertopic_available and claude_available:
    print('\n🔍 ANÁLISIS DE KEYWORDS:\n')
    print('='*80)

    # Obtener todas las keywords de ambos métodos
    bertopic_keywords = set()
    for arch in bertopic_results['archetypes']:
        bertopic_keywords.update(arch['keywords'])

    claude_keywords = set()
    for arch in claude_results['suggested_archetypes']:
        claude_keywords.update(arch['keywords'])

    # Calcular solapamiento
    common_keywords = bertopic_keywords & claude_keywords
    bertopic_only = bertopic_keywords - claude_keywords
    claude_only = claude_keywords - bertopic_keywords

    print(f"\nKeywords totales:")
    print(f"  - BERTopic: {len(bertopic_keywords)}")
    print(f"  - Claude: {len(claude_keywords)}")
    print(f"  - En común: {len(common_keywords)} ({len(common_keywords)/min(len(bertopic_keywords), len(claude_keywords))*100:.1f}%)")

    if len(common_keywords) > 0:
        print(f"\n✓ Keywords en común:")
        print(f"  {', '.join(sorted(list(common_keywords))[:15])}")

    print(f"\n💡 Interpretación:")
    if len(common_keywords) / min(len(bertopic_keywords), len(claude_keywords)) > 0.5:
        print("  ✅ Alto solapamiento: Ambos métodos identifican temas similares")
    elif len(common_keywords) / min(len(bertopic_keywords), len(claude_keywords)) > 0.25:
        print("  ⚠️  Solapamiento moderado: Algunos temas coinciden, otros son únicos")
    else:
        print("  ❌ Bajo solapamiento: Los métodos encuentran temas muy diferentes")
````

## 6. Visualización Comparativa

````python
if bertopic_available and claude_available:
    # Crear gráfico de barras comparativo
    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Arquetipos BERTopic', 'Arquetipos Claude'),
        specs=[[{'type': 'bar'}, {'type': 'bar'}]]
    )

    # BERTopic
    bertopic_names = [arch['name'] for arch in bertopic_results['archetypes']]
    bertopic_counts = [arch['thread_count'] for arch in bertopic_results['archetypes']]

    fig.add_trace(
        go.Bar(
            x=bertopic_counts,
            y=bertopic_names,
            orientation='h',
            marker=dict(color='#0045D7'),
            name='BERTopic'
        ),
        row=1, col=1
    )

    # Claude
    claude_names = [arch['nombre'] for arch in claude_results['suggested_archetypes']]
    claude_pcts = [arch.get('porcentaje_estimado', 0) for arch in claude_results['suggested_archetypes']]

    fig.add_trace(
        go.Bar(
            x=claude_pcts,
            y=claude_names,
            orientation='h',
            marker=dict(color='#DD0000'),
            name='Claude'
        ),
        row=1, col=2
    )

    fig.update_xaxes(title_text="Threads", row=1, col=1)
    fig.update_xaxes(title_text="% Estimado", row=1, col=2)

    fig.update_layout(
        height=600,
        showlegend=False,
        title_text="Distribución de Arquetipos: BERTopic vs Claude",
        font=dict(family="DM Sans, sans-serif")
    )

    fig.show()
````

## 8. Exportar Arquetipos Híbridos (Opcional)

````python
if bertopic_available and claude_available:
    print("\n💡 Tip: Puedes crear arquetipos híbridos manualmente")
    print("combinando lo mejor de ambos métodos:\n")
    print("1. Toma clusters de BERTopic con buen Silhouette Score")
    print("2. Usa nombres de Claude si son más descriptivos")
    print("3. Combina keywords de ambos")
    print("4. Usa descripciones de Claude")
    print("5. Valida con métricas de BERTopic")
    print("\n¿Quieres crear esta combinación ahora? (código comentado abajo)")
````

````python
# CÓDIGO PARA CREAR ARQUETIPOS HÍBRIDOS
# Descomenta y ajusta según necesites

hybrid_archetypes = []

# Ejemplo: Tomar arquetipos de BERTopic y mejorar nombres con Claude
for bert_arch in bertopic_results['archetypes'][:5]:  # Top 5 de BERTopic
    # Buscar arquetipo similar en Claude por keywords
    similar_claude = None
    for claude_arch in claude_results['suggested_archetypes']:
        # Comparar keywords
        common = set(bert_arch['keywords']) & set(claude_arch['keywords'])
        if len(common) >= 2:  # Si tienen al menos 2 keywords en común
            similar_claude = claude_arch
            break

    hybrid = {
        'name': similar_claude['nombre'] if similar_claude else bert_arch['name'],
        'description': similar_claude['descripcion'] if similar_claude else bert_arch['description'],
        'keywords': list(set(bert_arch['keywords'] + (similar_claude['keywords'] if similar_claude else []))),
        'priority': 100,
        'source': 'hybrid_bertopic_claude'
    }

    hybrid_archetypes.append(hybrid)

# Exportar
with open('hybrid_archetypes.json', 'w', encoding='utf-8') as f:
    json.dump(hybrid_archetypes, f, indent=2, ensure_ascii=False)

print(f'✓ Exportados {len(hybrid_archetypes)} arquetipos híbridos')
````
